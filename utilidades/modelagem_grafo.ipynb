```python
import pandas as pd
import networkx as nx

"""
Esse é o código que faz a modelagem de nossos dados como um grafo! :3
"""

df = pd.read_csv("../dados_processados/origem_destino_customedio.csv", sep=";", index_col=0)
df = df[df["mes"] == 1]

G = nx.from_pandas_edgelist(
    df,
    source="origem",
    target="destino",
    edge_attr=["custo_por_assento", "empresa"],
    create_using=nx.MultiDiGraph,
)
```

```python
list(G.edges(data=True))
```

Preparando o grafo para ser recebido pelo algoritmo de Yen:

```python
print(dict(G._adj["SBGL"]))
```
